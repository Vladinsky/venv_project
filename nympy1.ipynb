```python
import numpy as np 
arr2 = np.arange(16)
print("Shape of the arr2", arr2.shape)
arr2_copy = arr2.copy()
print("arr2",arr2)
print("arr2_copy",arr2_copy)
arr2 = arr2.reshape(4,4)
print("Reshaped arr2:\n",arr2)
print("arr2_copy (unchanged):",arr2_copy)
print("Number of dimensions in arr2:", arr2.ndim)
print("Rows in arr2:", arr2.shape[0])
print("Columns in arr2:", arr2.shape[1])
print("Total elements in arr2:", arr2.size)
print("Data type of arr2:", arr2.dtype)
percentile_50 = np.percentile(arr2, 50)
print("50th percentile of arr2:", percentile_50)
median = np.median(arr2)
print("Median of arr2 elements:", median)
```

```python
import matplotlib.pyplot as plt
import numpy as np

s = np.random.normal(loc=3, scale=4, size=100)

plt.hist(s, bins=15, color='skyblue', edgecolor='black')
plt.title("Histogram of Normally Distributed Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(alpha=0.3)
plt.show()
```
