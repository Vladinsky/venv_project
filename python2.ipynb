## Iterators While Loop
In your game, you need to create a special pyramid artifact for your characters. The shape of the pyramid should look like this:

**
To create this pyramid:

Use a while loop and print statement to generate each line. Each line should have one more * character than the line before it.

```python
def pyramid(n):
    i=0
    while i < n:
        print(" "*(n-i-1) + "*"*(2*i+1))
        i += 1

if __name__ == "__main__":
    n = int(input("Enter the number of rows: "))
    pyramid(n)
```
