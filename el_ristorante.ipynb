Creare un piccolo db di un ristorante con 10 taboli con posti diversi
Sito di prenotazioni


```python
database_tavoli = {"table1" : {"numero_posti" : 6, "prenotato" : False, "orario_prenotazione" : None},
                   "table2" : {"numero_posti" : 4, "prenotato" : False, "orario_prenotazione" : None},
                    "table3" : {"numero_posti" : 2, "prenotato" : False, "orario_prenotazione" : None},
                    "table4" : {"numero_posti" : 8, "prenotato" : False, "orario_prenotazione" : None},
                    "table5" : {"numero_posti" : 6, "prenotato" : False, "orario_prenotazione" : None},
                    "table6" : {"numero_posti" : 4, "prenotato" : False, "orario_prenotazione" : None},
                    "table7" : {"numero_posti" : 2, "prenotato" : False, "orario_prenotazione" : None},
                    "table8" : {"numero_posti" : 8, "prenotato" : False, "orario_prenotazione" : None},
                    "table9" : {"numero_posti" : 6, "prenotato" : False, "orario_prenotazione" : None},
                    "table10" : {"numero_posti" : 4, "prenotato" : False, "orario_prenotazione" : None}}
```

```python
def prenota_tavolo():
    print("Benvenuto nella prenotazione tavoli")
    print("I tavoli disponibili sono:")
    for tavolo in database_tavoli:
        if database_tavoli[tavolo]["prenotato"] == False:
            print(tavolo)
    tavolo_scelto = input("Quale tavolo vuoi prenotare? ")
    if database_tavoli[tavolo_scelto]["prenotato"] == False:
        numero_persone = int(input("Quante persone siete? "))
        if numero_persone <= database_tavoli[tavolo_scelto]["numero_posti"]:
            database_tavoli[tavolo_scelto]["prenotato"] = True
            database_tavoli[tavolo_scelto]["orario_prenotazione"] = input("A che ora vuoi prenotare il tavolo? ")
            print("Tavolo prenotato con successo")
        else:
            print("Il tavolo scelto non ha abbastanza posti")
    else:
        print("Il tavolo scelto è già prenotato")

def cancella_prenotazione():
    print("Benvenuto nella cancellazione prenotazione tavoli")
    print("I tavoli prenotati sono:")
    for tavolo in database_tavoli:
        if database_tavoli[tavolo]["prenotato"] == True:
            print(tavolo)
    tavolo_scelto = input("Quale tavolo vuoi cancellare la prenotazione? ")
    if database_tavoli[tavolo_scelto]["prenotato"] == True:
        database_tavoli[tavolo_scelto]["prenotato"] = False
        database_tavoli[tavolo_scelto]["orario_prenotazione"] = None
        print("Prenotazione cancellata con successo")
    else:
        print("Il tavolo scelto non è prenotato")

def visualizza_tavoli():
    print("Benvenuto nella visualizzazione tavoli")
    print("I tavoli disponibili sono:")
    for tavolo in database_tavoli:
        if database_tavoli[tavolo]["prenotato"] == False:
            print(tavolo)
    print("I tavoli prenotati sono:")
    for tavolo in database_tavoli:
        if database_tavoli[tavolo]["prenotato"] == True:
            print(tavolo)

def main():
    while True:
        print("Benvenuto nel ristorante")
        print("Cosa vuoi fare?")
        print("1. Prenota tavolo")
        print("2. Cancella prenotazione")
        print("3. Visualizza tavoli")
        print("4. Esci")
        scelta = input("Scelta: ")
        if scelta == "1":
            prenota_tavolo()
        elif scelta == "2":
            cancella_prenotazione()
        elif scelta == "3":
            visualizza_tavoli()
        elif scelta == "4":
            break
        else:
            print("Scelta non valida")

if __name__ == "__main__":
    main()
```
