crea una funzione che riporta il numero di occorrenze per ogni lettera di una stringa

```python
def count_letters_occurencies(string):
    dizionario = {}
    for letter in string:
        if letter in dizionario:
            dizionario[letter] += 1
        else:
            dizionario[letter] = 1
    return dizionario

print(count_letters_occurencies("ciao come va?"))
 
```
